1º - Importar bibliotecas, usaremos movipy 1.0.3 e numpy 1.26.0

```python
import moviepy.editor as mpe
import numpy as np
```

2º - Carregar ao video "projeto.mp4'

```python
video = mpe.VideoFileClip("video.mp4")
```

### Questão 1: Inverter a imagem, no eixo, a cada 20 segundo

1º - Criar função de inversão de imagem, para tal cortaremos o video a cada 20s e depois alteraremos o eixo, ao final concatenaremos tudo.

```python
def cortar_alternar_eixo(video, intervalo_segundos):
    duracao_total = int(video.duration)
    subclipes = []
    aplicar_espelhamento = False

    for inicio in range(0, duracao_total, intervalo_segundos):
        fim = min(inicio + intervalo_segundos, duracao_total)
        subclipe = video.subclip(inicio, fim)
        if aplicar_espelhamento:
            subclipe = subclipe.fl_image(lambda frame: np.fliplr(frame))
        subclipes.append(subclipe)
        aplicar_espelhamento = not aplicar_espelhamento
        
    return concatenate_videoclips(subclipes)
```

2º - Chamar função de inverter imagem a cada 20s, enviando como parâmetro o video que se quer fazer a edição.

```python
from moviepy.editor import concatenate_videoclips
video_processado = cortar_alternar_eixo(video, 20)
video_processado.write_videofile("video_eixo.mp4")
```

### Questão 2: Diminiur o som, de forma gradativa, a cada 30 segundos, de forma que os últimos 10 segundos fique em silêncio (sem som)


1º - Criar função de diminuir som, para tal cortaremos o video a cada 30s e depois multiplicaremos o som pelo fator de volume, ao final concatenaremos tudo.

```python
def edicao_volume(video, duracao, fator_volume):
    clipes = []
    for inicio in range(0, duracao - 10, 30):
        fim = min(inicio + 30, duracao - 10)
        fator = fator_volume ** ((inicio + 1) // 30)
        trecho = video.subclip(inicio, fim).fx(mpe.afx.volumex, fator)
        clipes.append(trecho)

    clipes.append(video.subclip(duracao - 10, duracao).fx(mpe.afx.volumex, 0.0))
    return concatenate_videoclips(clipes)
```

2º - Chamar função de edicao de volume, enviando como parâmetro o video, duracao do mesmo e fator de multiplicação do volume.

```python
video = mpe.VideoFileClip("video_eixo.mp4")
duracao = int(video.duration)
video_som = edicao_volume(video_processado, duracao, 0.60)
video_som.write_videofile("video_som.mp4")
```

### Questão 3: Após o primeiro minuto 60 segundos, fazer um corte para o segundo 80, por fim, insira os 20 segundos cortados no fim do video

1º - Criar função de cortar o video.

```python
def cortar_video(video, duracao, primeiro_corte, segundo_corte):
    trecho1 = video.subclip(0, primeiro_corte)
    trecho2 = video.subclip(primeiro_corte, segundo_corte)
    trecho3 = video.subclip(segundo_corte, duracao)
    return concatenate_videoclips([trecho1, trecho3, trecho2])
```

2º - Chamar função de edicao de volume, enviando como parâmetro o video, duracao do mesmo, momento do primeiro e segundo corte em segundos. 
Ao final salvar video final

```python
import moviepy.editor as mpe

video = mpe.VideoFileClip("video_som.mp4")
duracao = int(video.duration)

video_reordenado = cortar_video(video, duracao, 60, 80)
video_reordenado.write_videofile("video_final.mp4")
```
